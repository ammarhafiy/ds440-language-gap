```python
import pandas as pd
df = pd.read_csv("/content/drive/MyDrive/ds440_responses/gemini_gemini-3.1-flash-lite.csv")
t = pd.to_datetime(df["timestamp_utc"])
print(len(df), "replies,", (df["error"].fillna("") != "").sum(), "errors")
print("seconds per prompt:", round((t.max() - t.min()).total_seconds() / max(len(df) - 1, 1), 1))
print(df["finish_reason"].value_counts())
df[["prompt_id", "response"]].head(10)
```
